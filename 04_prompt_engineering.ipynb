# Chain Of thought

Chain of Thought (CoT) prompting encourages the model to think step-by-step before giving a final answer. It dramatically improves accuracy on reasoning tasks.

"Don't ask the model for the answer. Ask it to show its work."

```python
SYSTEM_PROMPT= """
You are an expert AI assistant in resolving user queries using chain of thought. 
You work on START, PLAN AND OUTPUT steps. 
You need to first PLAN what needs to be done. The PLAN can be multiple steps. 
Once you think enough PLAN has been done. finally you can give an OUTPUT.

Must follow instruction:
- strictly follow the Json output format
- only run one step at a time
-  The sequence of steps is START(where user give an input), PLAN (that can be multiple itmes) and finally
OUTPUT (which is going to be displayed to the user)

OUTPUT Json Format: 
{"step": "START"| "PLAN"| "OUTPUT", "content": "<string>"}

Example: 
START: Hey, Can you solve 2 + 3 * 5 / 10
PLAN: {"step": "PLAN", "content": "Seems user likes to solve a math problem"}
PLAN: {"step": "PLAN", "content": "Looks like its a combination of arithmetic problem which can be solved using BODMAS methon"}
PLAN: {"step": "PLAN", "content": "Yes, The BODMAS is correct way of solving it"}
PLAN: {"step": "PLAN", "content": "first we my divide by 5 by 10 which is 0.5"}
PLAN: {"step": "PLAN", "content": "next we my multiply by 0.5 by 3 which is 1.5"}
PLAN: {"step": "PLAN", "content": "then we my add by 1.5 to 2 which is 3.5"}
PLAN: {"step": "PLAN", "content": "Great, We finally have an answer which is 3.5"}
OUTPUT: {"step": "OUTPUT", "content": "The answer is 3.5"}
"""
```

```python
import json

messages = [
    {"role": "system", "content": SYSTEM_PROMPT},
    {"role": "user", "content": "Give me a python code to add n numbers"},
    {"role": "assistant", "content": json.dumps({"step": "START", "content": "User wants a Python code snippet to add n numbers."})},
    {"role": "assistant", "content": json.dumps({"step": "PLAN", "content": "We need to provide a Python function that takes a number 'n' and then adds 'n' numbers."})},
    {"role": "assistant", "content": json.dumps({"step": "PLAN", "content": "The function should accept multiple numbers either as input parameters or by taking input from the user."})},
    {"role": "assistant", "content": json.dumps({"step": "PLAN", "content": "We can write a function that takes a list of numbers and returns their sum."})},
    {"role": "assistant", "content": json.dumps({"step": "PLAN", "content": "Alternatively, we can write code that interactively takes 'n' numbers as input from the user and computes their sum."})},
    {"role": "assistant", "content": json.dumps({"step": "PLAN", "content": "Combining the ideas, we will provide a function that takes 'n' numbers as input and returns the sum, and also demonstrate code that uses this function."})},
    ]
repsonse= openai.chat.completions.create(
    model="gpt-4.1-mini", 
    response_format={"type": "json_object"},
    messages=messages
    )
print(repsonse.choices[0].message.content)

```

```python
import json

print("\n\n\n\n\n")

message_history = [{"role": "system", "content": SYSTEM_PROMPT}]

user_query = input("👉🏻")
message_history.append({"role": "user", "content": user_query})

while True:
    response = openai.chat.completions.create(
        model="gpt-4.1-mini",
        response_format={"type":"json_object"},
        messages=message_history
        )
    raw_result = response.choices[0].message.content
    message_history.append({"role": "assistant", "content": raw_result})

    parsed_result = json.loads(raw_result)

    if parsed_result.get("step") == "START":
        print(f"🔥: {parsed_result.get('content')}")
        continue
    elif parsed_result.get("step") == "PLAN":
        print(f"🔥: {parsed_result.get('content')}")
        continue
    elif parsed_result.get("step") == "OUTPUT":
        print(f"🤖: {parsed_result.get('content')}")
        break

    print("\n\n\n\n\n")
```

# Persona Based Prompting

```python
SYSTEM_PROMPT ="""
        You are Sherlock Holmes.

You possess extraordinary powers of observation and deduction.
Never answer immediately.

First, examine every clue.
State your observations.
Form multiple hypotheses.
Eliminate weaker possibilities.
Then reveal your final deduction.

Speak in the elegant, confident style of Sherlock Holmes.

"""
messages = [
    {"role": "system", "content": SYSTEM_PROMPT},
    {"role": "user", "content": "My laptop battery lasts only 40 minutes. What could be wrong?"}
]


response = openai.chat.completions.create(
        model="gpt-4.1-mini",
        messages=messages
        )

print(response.choices[0].message.content)
```
